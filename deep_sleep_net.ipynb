```python
import numpy as np

from utils.file_handler import get_cassette_telemetry_pairs, combine_paths
from utils.edf_handler import generate_epochs_from_edf


origin_folder_path = 'sleep-edf-database-expanded-1.0.0'
folder_names = ['sleep-cassette', 'sleep-telemetry']
epoch_length = 30  # độ dài mỗi epoch (giây)

# Lấy danh sách các cặp file
filename_in_pair_all_folders = get_cassette_telemetry_pairs(
    origin_folder_path, folder_names[0], folder_names[1]
)
print(f"Found {len(filename_in_pair_all_folders)} pairs of files.")
print(f"Example pair: {filename_in_pair_all_folders[0]}")
```

```python
try:
    # Duyệt qua từng folder (thường list này có 2 phần tử: cassette và telemetry)
    for folder_idx, folder_content in enumerate(filename_in_pair_all_folders):
        current_subfolder = folder_names[0] if folder_idx == 0 else folder_names[1]
        print(f"\n=== ĐANG XỬ LÝ FOLDER: {folder_names[folder_idx].upper()} ===")

        # folder_content lúc này là dict kiểu: {'SC4001': {...}, 'SC4002': {...}}
        # Chúng ta duyệt qua từng ID bệnh nhân (patient_id)
        for patient_id, pair in folder_content.items():
            psg_path = combine_paths(origin_folder_path, current_subfolder, pair['psg'])
            hyp_path = combine_paths(origin_folder_path, current_subfolder, pair['hypnogram'])
            
            print(f"Đang xử lý cặp: {patient_id} | File: {pair['psg']}")

            # Tạo Epochs
            epochs = generate_epochs_from_edf(psg_path, hyp_path)
            
            # Save as npz
            save_path = f"processed_data/raw_data/{patient_id}_{current_subfolder}.npz"
            np.savez_compressed(save_path, data=epochs.get_data(), labels=epochs.events[:, -1])
            print(f"Saved processed data to: {save_path}")

            # 5. Dọn dẹp RAM ngay lập tức
            del epochs

except Exception as e:
    print(f"Lỗi khi đang xử lý: {e}")
```

```python
import os
import numpy as np
import torch

from model_development.deepsleepnet import DeepSleepNet

save_path = "processed_data/raw_data/"

def load_all_features_numpy(data_dir, folder_names=folder_names):
    print("Đang tải và gộp các file .npz...")
    all_data = []
    all_labels = []

    if folder_names:
        print(f"Chỉ tải dữ liệu từ các folder: {folder_names}")
    else:
        print("Tải dữ liệu từ tất cả các folder con trong thư mục gốc.")
    
    for file_name in os.listdir(data_dir):
        if file_name.endswith('.npz'):
            file_path = os.path.join(data_dir, file_name)
            with np.load(file_path) as data:
                # Đọc trực tiếp mảng numpy
                all_data.append(data['data'])
                all_labels.append(data['labels'])
                
    # Gộp tất cả các file lại theo trục hàng (Axis 0) thành 1 mảng lớn duy nhất
    # Cách này nhanh và chuẩn hơn dùng Pandas DataFrame rất nhiều đối với dữ liệu thô lớn
    X = np.concatenate(all_data, axis=0)
    y = np.concatenate(all_labels, axis=0)
    return X, y

try:
    print("\n=== BẮT ĐẦU TRAINING DEEPSLEEPNET ===")
    
    X, y = load_all_features_numpy(save_path, folder_names=[folder_names[0]])  # Chỉ tải dữ liệu từ folder 'sleep-cassette'
    print(f"Tổng số dữ liệu tải về: X shape = {X.shape} | y shape = {y.shape}")
    
    split_idx = int(0.8 * len(X))
    X_train, y_train = X[:split_idx], y[:split_idx]
    X_val, y_val = X[split_idx:], y[split_idx:]
    
    if torch.cuda.is_available():
        device = "cuda"
    elif torch.backends.mps.is_available():
        device = "mps"
    else:
        device = "cpu"
        
    # Khởi tạo mô hình và đẩy lên thiết bị (CPU/CUDA/MPS)
    model = DeepSleepNet(num_classes=5).to(device)

    trained_model = model.fit(
        X_train, y_train, X_val, y_val, 
        epochs=5, batch_size=16, device=device
    )

    # Load features tu sleep_telemetry
    X_test, y_test = load_all_features_numpy(save_path, folder_names=[folder_names[1]])  # Chỉ tải dữ liệu từ folder 'sleep-telemetry'
    print(f"Tổng số dữ liệu test tải về: X_test shape = {X_test.shape} | y_test shape = {y_test.shape}")

    # Gọi hàm evaluation sau khi huấn luyện xong
    val_acc = trained_model.evaluate(X_test, y_test, batch_size=16, device=device)
    print(f"Validation Accuracy sau khi training: {val_acc:.4f}")
    

    print("Đã hoàn thành training DeepSleepNet và lưu mô hình thành công!")

except Exception as e:
    print(f"Lỗi khi đang train DeepSleepNet: {e}")
```
